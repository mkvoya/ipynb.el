# ipynb.el demo

Press **RET** to edit a cell, **K** to start its kernel, and **L** to start cross-cell LSP.

## New tools

| Key | Action |
| --- | --- |
| `v` | Inspect objects in the current code cell |
| `P` | Toggle Markdown source / preview |
| `C-c l d` | Jump to a definition in another cell |
| `C-c l n` | Rename across cells |

- Markdown is rendered inside Emacs.
- Cell separators follow the window width.
